Problem1: Largest Populus:

```python
def largest_population(population):
    largest = float('-inf')
    for a in population:
        if largest < a:
            largest = a
    return largest
```

```python
pop = [-5, -2, -9]
large = largest_population(pop)
print(large)
```

Problem2: Most Populus

```python
def most_populus(counties):
    most = float('-inf')
    county = None
    a = len(counties)
    k = list(counties.keys())
    l = list(counties.values())
    for i in range(a):
        if most < l[i]:
            most = l[i]
            county = k[i]
    return county
```

```python
c = {"Madison": 110000,"Hinds": 227000, "Rankin": 157000}
mp = most_populus(c)
print(mp)
```

```python
def most_populous(counties):
    most = float('-inf')
    county = None
    for name,pop in counties.items():
        if most < pop:
            most = pop
            county = name
    return county
```

```python
c = {"Madison": 110000,"Hinds": 227000, "Rankin": 157000}
mp = most_populous(c)
print(mp)
```

Problem4: Has Duplicate:

```python
def has_duplicate(codes):
    di = {}
    for code in codes:
        di[code] = di.get(code,0) + 1
    for v in di.values():
        if v > 1:
            return True
    return False
```

```python
#codes = ["28049", "48201", "28089"]           
#codes = ["28049", "48201", "28049", "01073"] 
#codes = []
codes = ["s","r","q","p","p"]
fu = has_duplicate(codes)
print(fu)
```

```python
def has_duplicate(codes):
    s = set()
    for code in codes:
        if code in s:
            return True
        s.add(code)
    return False
```

```python
codes = ["28049", "48201", "28089"]           
#codes = ["28049", "48201", "28049", "01073"] 
#codes = []
#codes = ["s","r","q","p","p"]
fu = has_duplicate(codes)
print(fu)
```

Problem 5: First unique.

```python
def first_unique(values):
    d = {}
    for v in values:
        d[v] = d.get(v,0)+1
    for key,value in d.items():
        if value == 1:
            return key
    return None
```

```python
def first_unique(values):
    d = {}
    for v in values:
        d[v] = d.get(v,0)+1
    for name in values:
        if d[name] == 1:
            return name

    return None
```

Problem 6: filter counties by threshold.

```python
def big_counties(counties,threshold):
    result = []
    for key,value in counties.items():
        if value > threshold:
            result.append(key)
    return result
```

Problem7: Group counties by state.

```python
def group_by_state(pairs):
    result = {}
    for county,state in pairs:
        if state in result:
            result[state].append(county)
        else:
            result[state] = [county] 
    return result
```
