```python
def largest_population(population):
    largest = float('-inf')
    for a in population:
        if largest < a:
            largest = a
    return largest
```

Problem2: Most Populus

```python
def most_populus(counties):
    most = float('-inf')
    county = None
    a = len(counties)
    k = list(counties.keys())
    l = list(counties.values())
    for i in range(a):
        if most < l[i]:
            most = l[i]
            county = k[i]
    return county
```

```python
c = {"Madison": 110000,"Hinds": 227000, "Rankin": 157000}
mp = most_populus(c)
print(mp)
```

```python
def most_populous(counties):
    most = float('-inf')
    county = None
    for name,pop in counties.items():
        if most < pop:
            most = pop
            county = name
    return county
```

```python
c = {"Madison": 110000,"Hinds": 227000, "Rankin": 157000}
mp = most_populous(c)
print(mp)
```

Problem4: Has Duplicate:

```python
def has_duplicate(codes):
    di = {}
    for code in codes:
        di[code] = di.get(code,0) + 1
    for v in di.values():
        if v > 1:
            return True
    return False
```

Problem9: Pair of the Target.

```python
def pair_for_target(rows, target):
    length = len(rows)
    for i in range(length):
        for j in range(i+1,length):
            name_i,population_i = rows[i]
            name_j,population_j =  rows[j]
            if target == (population_i + population_j):
                return (name_i,name_j)
    return None
```

```python
def pair_for_target(rows, target):
    seen = {}
    for name,pop in rows:
        needed = target - pop
        if needed in seen:
            return (seen[needed],name)
        seen[pop] = name
    return None
```

```python
rows = [("Hinds", 227000), ("Madison", 110000), ("Rankin", 157000), ("Lee", 174000)]

#pair_for_target(rows, 267000)   #→ ("Madison", "Rankin")    # 110000 + 157000
#pair_for_target(rows, 401000)   #→ ("Hinds", "Lee")         # 227000 + 174000
#pair_for_target(rows, 999)      #→ None
ans = pair_for_target([("Kent", 500)], 1000) #→ None              # can't use Kent twice
print(ans)
```

Problem12: Fixing the case of the name.

```python
def fix_case(name):
    result = []
    for i in range(len(name)):
        if i == 0 or name[i-1] == " ":
            result.append(name[i].upper())
        else:
            result.append(name[i].lower())
    return "".join(result)
            
```

Problem16: Finding duplicates.

```python
def dedupe(names):
    result = []
    seen = set()
    for name in names:
        if name not in seen:
            seen.add(name)
            result.append(name) 
    return result
```

Problem17: Same codes, Different order.

```python
def same_codes(a, b):
    a_dic = {}
    b_dic = {}
    if len(a) != len(b):
        return False

    for code in a:
        a_dic[code] = a_dic.get(code,0)+1
    
    for code in b:
        b_dic[code] = b_dic.get(code,0)+1
    return a_dic == b_dic
```

Problem19: Second largest 

```python
def second_largest(pops):
    first = float('-inf')
    second = float('-inf')
    for pop in pops:
        if  pop > first:
            second = first
            first = pop
        elif pop > second and pop < first:
            second = pop
    if second == float('-inf'):
        return None
    return second

```

Problem20: Compress Status Flag.

```python
def compress(flags):
    count = 1
    result = ""
    if flags == "":
        return "Empty string"
    for i in range(1,len(flags)):
        if flags[i] == flags[i-1]: 
            count += 1
        else:
            result += flags[i-1]+str(count) 
            count = 1
    result+= flags[-1]+str(count)
    return result
```

Problem21: Palindrome.

```python
def is_mirror(code):
    left = 0
    right = len(code) - 1
    while left < right:
        if code[left] != code[right] :
            return False
        left += 1
        right -=1
    return True

```

Problem22: Valid Palindrome.

```python
def is_palindrome(s):
    left = 0
    right = len(s) - 1
    while left < right:
        while not s[left].isalnum() and left < right:
            left +=1
        while not s[right].isalnum() and left < right:
            right -=1
        if s[left].lower() != s[right].lower():
            return False
        left +=1
        right -=1
    return True
```

Problem23: Two sum on a sorted list.

```python
def pair_sorted(pops, target):
    left = 0
    right = len(pops) - 1
    while left < right:
        total = pops[left] + pops[right]
        if total > target:
            right -= 1
        elif total < target:
            left +=1
        else:
            return [left,right]
    return None
```

Problem24: Merge Sorted Lists.

```python
def merge_sorted(a, b):
    i=0
    j=0
    result = []
    while i < len(a) and j < len(b):

        if a[i] < b[j]:
            result.append(a[i])
            i+=1
        else:
            result.append(b[j])
            j+=1
    while i < len(a):
        result.append(a[i])
        i +=1
    while j < len(b):
        result.append(b[j])
        j +=1
    return result
```

Problem26: Counties in both files.

```python
def in_both(a, b):
    b_lookup = set(b)
    seen = set()
    result = []
    for name in a:
        if name in b_lookup and name not in seen:
            result.append(name)
            seen.add(name)
    return result
```

Problem28:Best month strech.

```python
def best_window(counts, k):
    if k > len(counts):
        return None
    best = 0
    total = 0
    nums = counts[0:k]
    for num in nums:
        total += num 
    best = total
    for i in range(k,len(counts)):
        total += counts[i] - counts[i-k]
        if total > best:
            best = total
    return best
    
```

```python
#ans = best_window([3, 5, 2, 6, 1, 4], 3)   #→ 13    # 5 + 2 + 6
#ans = best_window([3, 5, 2, 6, 1, 4], 1)   #→ 6
ans = best_window([4, 4, 4], 3)            #→ 12
#ans = best_window([1, 2], 3)               #→ None
print(ans)
```

Problem29: Longest strech under a limit.

```python
def longest_under(counts, limit):
    best = 0
    left = 0
    total = 0
    for right in range(len(counts)):
        total += counts[right]
        while total > limit:
            total -= counts[left]
            left +=1
        length = right -  left + 1
        if length > best:
            best = length
    return best

```

```python
#ans = longest_under([2, 1, 3, 1, 1, 4], 5)  # → 3     # [3, 1, 1] = 5
#ans = longest_under([1, 1, 1, 1], 10)        #→ 4     # the whole list fits
#ans = longest_under([6, 7, 8], 5)            #→ 0     # no single month fits
ans = longest_under([], 5)                   #→ 0
print(ans)
```

Problem30: Longest stretch with no repeated values.

```python
def longest_unique_run(states):
    seen = set()
    left = 0
    best = 0
    for right in range(len(states)):
        while states[right] in seen:
            seen.remove(states[left])
            left +=1
        seen.add(states[right])
        length = right - left +1
        if length > best:
            best =  length
    return best
```
