Problem1: Largest Populus:

```python
def largest_population(population):
    largest = float('-inf')
    for a in population:
        if largest < a:
            largest = a
    return largest
```

```python
pop = [-5, -2, -9]
large = largest_population(pop)
print(large)
```

Problem2: Most Populus

```python
def most_populus(counties):
    most = float('-inf')
    county = None
    a = len(counties)
    k = list(counties.keys())
    l = list(counties.values())
    for i in range(a):
        if most < l[i]:
            most = l[i]
            county = k[i]
    return county
```

```python
c = {"Madison": 110000,"Hinds": 227000, "Rankin": 157000}
mp = most_populus(c)
print(mp)
```

```python
def most_populous(counties):
    most = float('-inf')
    county = None
    for name,pop in counties.items():
        if most < pop:
            most = pop
            county = name
    return county
```

```python
c = {"Madison": 110000,"Hinds": 227000, "Rankin": 157000}
mp = most_populous(c)
print(mp)
```

Problem4: Has Duplicate:

```python
def has_duplicate(codes):
    di = {}
    for code in codes:
        di[code] = di.get(code,0) + 1
    for v in di.values():
        if v > 1:
            return True
    return False
```

```python
#codes = ["28049", "48201", "28089"]           
#codes = ["28049", "48201", "28049", "01073"] 
#codes = []
codes = ["s","r","q","p","p"]
fu = has_duplicate(codes)
print(fu)
```

```python
def has_duplicate(codes):
    s = set()
    for code in codes:
        if code in s:
            return True
        s.add(code)
    return False
```

Problem 5: First unique.

```python
def first_unique(values):
    d = {}
    for v in values:
        d[v] = d.get(v,0)+1
    for key,value in d.items():
        if value == 1:
            return key
    return None
```

Problem7: Group counties by state.

```python
def group_by_state(pairs):
    result = {}
    for county,state in pairs:
        if state in result:
            result[state].append(county)
        else:
            result[state] = [county] 
    return result
```

Problem9: Pair of the Target.

```python
def pair_for_target(rows, target):
    length = len(rows)
    for i in range(length):
        for j in range(i+1,length):
            name_i,population_i = rows[i]
            name_j,population_j =  rows[j]
            if target == (population_i + population_j):
                return (name_i,name_j)
    return None
```

```python
rows = [("Hinds", 227000), ("Madison", 110000), ("Rankin", 157000), ("Lee", 174000)]

#pair_for_target(rows, 267000)   #→ ("Madison", "Rankin")    # 110000 + 157000
#pair_for_target(rows, 401000)   #→ ("Hinds", "Lee")         # 227000 + 174000
#pair_for_target(rows, 999)      #→ None
ans = pair_for_target([("Kent", 500)], 1000) #→ None              # can't use Kent twice
print(ans)
```

```python
def pair_for_target(rows, target):
    seen = {}
    for name,pop in rows:
        needed = target - pop
        if needed in seen:
            return (seen[needed],name)
        seen[pop] = name
    return None
```

```python
rows = [("Hinds", 227000), ("Madison", 110000), ("Rankin", 157000), ("Lee", 174000)]

#pair_for_target(rows, 267000)   #→ ("Madison", "Rankin")    # 110000 + 157000
#pair_for_target(rows, 401000)   #→ ("Hinds", "Lee")         # 227000 + 174000
#pair_for_target(rows, 999)      #→ None
ans = pair_for_target([("Kent", 500)], 1000) #→ None              # can't use Kent twice
print(ans)
```

Problem10: Most Populus State.

```python
def biggest_state(rows):
    d = {}
    result = None
    largest = float('-inf')
    for county,state,population in rows:
        d[state] = d.get(state,0)+population
        if d[state] > largest:
            largest = d[state]
            result = state
    return result
```

Problem11: Clean a FIPS code.

```python
def clean_fips(raw):
    result = ""
    for ch in raw:
        if ch.isdigit():
            result += ch
    return result
```

Problem12: Fixing the case of the name.

```python
def fix_case(name):
    result = []
    for i in range(len(name)):
        if i == 0 or name[i-1] == " ":
            result.append(name[i].upper())
        else:
            result.append(name[i].lower())
    return "".join(result)
            
```

```python
ans = fix_case("hinds county")  # → "Hinds County"
#ans = fix_case("DE SOTO")        #→ "De Soto"
#ans = fix_case("rankin")         #→ "Rankin"
#ans = fix_case("mcCLAIN")        #→ "Mcclain"
#ans = fix_case("")               #→ ""
print(ans)
```

Problem13: Change in yearly population.

```python
def yearly_changes(pops):
    result = []
    for i in range(0,len(pops)-1):
        diff = pops[i+1] - pops[i]
        result.append(diff)
    return result
```

Probelm14: longest growth in a list.

```python
def longest_growth(pops):
    current = 0
    result = 0
    for i in range(0,len(pops)-1):
        if pops[i] < pops[i+1]:
            current += 1
        else:
            current = 0
    if current > result:
            result = current
    return result
```
